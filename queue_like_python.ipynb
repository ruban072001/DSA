## QUEUE

- Queue is a collection of items that follows FIFO ==> First In First Out

##### Example in Real Life

- People Standing in line at ticket counter
- The first person who comes is the first to be served

```python
class queue:
    
    def __init__(self):
        self.items = []
        
    def enqueue(self, item):
        self.items.append(item)    # Add the item in rear / front
        
    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0) # Remove the item from the front
        return "Empty Queue" 
    
    def peek(self):
        if not self.is_empty():
            return self.items[0]  # Returnin the front element without removing it.
        return "Empty Queue"
    
    def is_empty(self):    
        return len(self.items) == 0   # Check if queue is empty
    
    def size(self):
        return len(self.items)        # Return the number of items in queue
```

```python
q = queue()
```

```python
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)
q.enqueue(60)
```
