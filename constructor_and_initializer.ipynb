A constructor is a special method that creates a new object in memory. In Python, __new__ is the actual constructor.

An initializer is a special method that sets up or initializes the object after it has been created. In Python, __init__ is the initializer.



1. `__new__` → makes the object (constructor, memory allocation).

2. `__init__` → sets up the object (initializer).

3. Using `super().__new__` → tells Python: "Okay, I added my custom step, now let the parent (object) actually build it."

```python
class Person:
    def __init__(self, name, age):
        print("Inside __init__")
        self.name = name 
        self.age = age
        self.is_adult = age >= 18

p = Person("Ruban", 30)
print(p.name, p.age, p.is_adult)
```

1. When we call this `class Person()`, Python automatically call `obj.__new__(cls)` to allocate memory for this object. 
2. p = Person("Ruban", 30) → Python allocates memory for the object automatically (via default __new__)

3. `__init__` is called automatically with self as the newly created object (Here self refering to the object that we have created now.)

4. Inside `__init__:`

    - self.name = name (sets the name attribute.)
    - self.age = age (sets the age attribute.)
    - self.is_adult = age >= 18 (sets another attribute based on age)

5. The object `p` now has all attributes initialized, ready to use. And we have the final output

```python
class MyInt(int):
    def __new__(cls, value):
        print("Inside __new__")
        # Create the object using parent class int.__new__
        return super().__new__(cls, value + 10)

n = MyInt(5)
print(n)
```

1. `n = MyInt(5)` triggers `__new__` first. (It triggers python default object creation)

2. Inside `__new__`:

    - `super().__new__(cls, value + 10)` calls `int.__new__` to actually create the integer object. This override the default object creation.
    - Here, we added 10 to the value, so 5 → 15

3. KEY Points:

    - You need to control how the object is created.
    - You work with immutable types (like int, str, tuple).
    - You implement singletons or caching (one instance reused).
    

| Feature         | `__init__`                        | `__new__`                                             |
| --------------- | --------------------------------- | ----------------------------------------------------- |
| Purpose         | Initialize instance attributes    | Create/construct the object                           |
| Called after    | Object is created                 | Called before object is created                       |
| Return value    | Must return `None`                | Must return the new object                            |
| Use cases       | Set attributes, validation, setup | Immutable objects, singletons, custom object creation |
| Usually needed? | ✅ Yes, almost always              | ❌ Rarely, only special cases                          |


```python
class A:
    def __new__(cls, *args, **kwargs):
        print("My custom __new__")
        return super().__new__(cls)   # still call object.__new__
a = A()
```

1. When you do `a = A()`, Python calls `A.__new__(A)`.

2. Since you overrode `__new__`, your custom code runs → it prints `"My custom __new__"`.

3. Then you call `super().__new__(cls)` → this goes up the inheritance chain (for A, the parent is object) and calls `object.__new__(A)`.

4. `object.__new__` actually allocates the memory and returns the new object of `type A`.

5. That new object is then passed to `__init__` (if defined).

Super() tells python to `“Do my custom stuff first, but then let the parent (object) do the actual object creation.”`
