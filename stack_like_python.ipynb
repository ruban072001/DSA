## Stack Implementation in Python

A stack is a data structure same like list[], tuple(), set{} to store the collection of items in it.

A stack is a collection of item that follow the LIFO rule: (Last In First Out).

We Push one element at a time. And pop one element at a time.

#### Example in real life:

    - Stack of plates 🍽️ → You put plates one on top of another.
    - When you need one, you take the top plate first, not the bottom.

```python
class Stack:
    def __init__(self):                      # Initialize an empty stack
        self.items = []
        
    def push(self, item):                    # Add an item to the top of the stack
        self.items.append(item)
        
    def pop(self):                           # Remove and return the top item of the stack
        if not self.is_empty():
            return self.items.pop()
        return None
    
    def peek(self):                           # Return the top item of the stack without removing it
        if not self.is_empty():
            return self.items[-1]
        return None
    
    def is_empty(self):                       # Check if the stack is empty
        return len(self.items) == 0
    
    def size(self):                            # Return the number of items in the stack
        return len(self.items)
        
```

```python
def check_balance(expression):
    stack = Stack()
    symbols = {"(", "{", "["}     # In set and dict the look-up is always O(1), which is constant and faster then normal string look-up like "({["
    map_symbols = {
        ")": "(",
        "}": "{",
        "]": "["
    }
    
    for char in expression:
        if char in symbols:
            stack.push(char)
            
        elif char in map_symbols:
            if stack.is_empty() or stack.pop() != map_symbols[char]:
                return False
                
    return stack.is_empty()
```

```python
check_balance("{)}")
```

```python
class Editor:
    
    def __init__(self):
        self.text = ""
        self.undo_text = []
        self.redo_text = []
        
    def write(self, text):
        self.undo_text.append(self.text)
        self.text += text
        self.redo_text.clear()
        
    def undo(self):
        if self.undo_text:
            self.text = self.undo_text.pop()
            self.redo_text.append(self.text)
            return self.text
        else:
            return ""
            
    def redo(self):
        if self.redo_text:
            self.text = self.redo_text.pop()
            self.undo_text.append(self.text)
            return self.text
        else:
            return self.text
```

```python
edit = Editor()
```

```python
edit.write("I")
edit.write(" am")
edit.write(" ruban")
edit.write(". ")
edit.write("Super")
edit.write(" I")
edit.write(" Think")
```
